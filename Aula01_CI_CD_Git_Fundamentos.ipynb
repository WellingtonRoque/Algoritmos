# 📌 Antes do Git: o Problema

Imagine o seguinte cenário:

Você está desenvolvendo um sistema.  
Salva como:

- sistema.py
- sistema_novo.py
- sistema_final.py
- sistema_final_agora_sim.py

Agora imagine isso em equipe.

Sem controle de versão:

- Não sabemos quem alterou o código
- Podemos sobrescrever código de outro colega
- Não temos histórico
- Não conseguimos voltar versões anteriores

👉 Esse problema é o que o Git resolve.


# 💡 O que é Git?

Git é um Sistema de Controle de Versão Distribuído (DVCS).

Ele permite:

✔ Registrar alterações no código  
✔ Trabalhar em equipe  
✔ Criar versões organizadas  
✔ Recuperar estados anteriores  
✔ Integrar com plataformas como GitHub  

---

## 🔗 E onde entra CI/CD?

A Integração Contínua depende de:

- Código versionado
- Histórico organizado
- Commits frequentes
- Controle estruturado de alterações

👉 Sem Git, não existe CI/CD.


# ⚙ Como o Git funciona?

O Git trabalha com três áreas principais:

1️⃣ Working Directory → Onde você altera arquivos  
2️⃣ Stage (Index) → Onde você prepara o que será salvo  
3️⃣ Commit → Registro permanente da alteração  

Fluxo básico:

Arquivo → git add → git commit → Histórico

Cada commit é como uma "foto" do projeto naquele momento.

# 📦 O que fez o git init?

O comando:

git init

Criou uma pasta oculta chamada .git

Essa pasta contém:

- Histórico
- Configurações
- Controle de versões

Agora nosso projeto está sob controle do Git.


# 🌐 O que é Flask?

Flask é um microframework para desenvolvimento de APIs e aplicações web em Python.

Ele permite:

- Criar rotas
- Responder requisições HTTP
- Retornar JSON
- Construir APIs modernas

Nossa API será simples, mas servirá como base para CI/CD.


```python
with open("app.py", "w") as f:
    f.write("""
from flask import Flask, jsonify

app = Flask(__name__)

@app.route("/")
def home():
    return jsonify({"message": "API CI/CD funcionando!"})

@app.route("/status")
def status():
    return jsonify({"status": "ok"})

if __name__ == "__main__":
    app.run(debug=True)
""")
```

# 🧠 Entendendo o código

@app.route("/")  
Define a rota principal da API.

jsonify(...)  
Retorna resposta no formato JSON.

app.run(debug=True)  
Inicia o servidor local para testes.

Essa aplicação será evoluída ao longo do semestre.


# 📦 O que é requirements.txt?

# 🔎 Verificando o estado do repositório

O comando git status mostra:

- Arquivos novos
- Arquivos modificados
- Arquivos prontos para commit


```python
git status
```

# ➕ O que é git add?

O git add move arquivos da Working Directory para o Stage.

É como dizer:

"Estou preparando essas alterações para serem registradas."


```python
git add .
```

# 💾 O que é um commit?

Um commit é um registro permanente no histórico.

Boas práticas de mensagens:

- feat: nova funcionalidade
- fix: correção
- docs: documentação
- chore: ajustes técnicos

Mensagens claras facilitam CI/CD e auditoria.


```python
git commit -m "feat: estrutura inicial da API Flask"
```

# 📜 Visualizando o histórico

git log --oneline

Mostra todos os commits realizados.


```python
git log --oneline
```

# 🔗 Por que isso importa para CI/CD?

Toda pipeline de CI funciona baseada em:

✔ Eventos de commit  
✔ Eventos de push  
✔ Mudanças no repositório  

Quando fizermos nossa primeira automação,
ela será acionada por um commit.

👉 O commit é o gatilho da integração contínua.
